```python
from openai import OpenAI

endpoint = "https://burinakeai-resource.services.ai.azure.com/openai/v1"
deployment_name = "Burinake-vlm"
api_key = API_KEY

client = OpenAI(
    base_url=endpoint,
    api_key=api_key
)

response = client.responses.create(
    model=deployment_name,
    input="What is the capital of France?",
)

print(f"answer: {response.output[1]}")
```
