```python
from ultralytics import YOLO

model = YOLO('./runs/detect/fire_smoke_rtdetr_aug-3/weights/best.pt')

print("🚀 기존 가중치를 성공적으로 불러왔습니다. 소량 데이터 추가 학습을 시작합니다!")

# 2. 소량 데이터 파인튜닝 특화 설정으로 학습 실행
results = model.train(
    data='../dataset/new_small_data.yaml', # 🌟 직접 라벨링한 소량의 데이터셋
    epochs=50,          # 과적합 방지를 위해 Early Stopping(patience)을 믿고 넉넉히 줍니다.
    batch=4,            
    imgsz=640,
    name='rtdetr_finetuned_small',
    
    # 🌟 핵심 전략 1: 극도로 낮은 학습률 (Low Learning Rate)
    # 기존 지식을 잊지 않도록 보폭을 아주 작게 설정합니다.
    lr0=0.00005,        # 기본 파인튜닝(0.0001)보다 더 낮춤
    lrf=0.01,           
    
    # 🌟 핵심 전략 2: 특징 추출기 얼리기 (Freeze)
    # RT-DETR 모델의 앞단 10개 레이어를 얼려서(학습 안 함) 기존의 범용적인 특징 인식 능력 보존
    freeze=10,
    
    # 🌟 핵심 전략 3: 조기 종료 (Early Stopping)
    # 10 Epoch 동안 검증 성능(mAP)이 개선되지 않으면 과적합으로 간주하고 스스로 멈춤
    patience=10,
    
    # 🌟 핵심 전략 4: 데이터 증강 (Augmentation) 강화
    # 적은 데이터를 뻥튀기하여 다양한 상황에 대비하게 함
    hsv_h=0.015,  # 색상 변화
    hsv_s=0.7,    # 채도 변화
    hsv_v=0.4,    # 명도 변화
    degrees=10.0, # 회전
    translate=0.1, # 이동
    scale=0.5,    # 크기 조절
    fliplr=0.5,   # 좌우 반전 확률 50%
    mosaic=1.0    # 4장의 이미지를 하나로 합치는 모자이크 기법 (아주 강력함)
)

print("✅ 추가 학습이 완료되었습니다. 더 똑똑해진 모델을 확인해보세요!")
```
