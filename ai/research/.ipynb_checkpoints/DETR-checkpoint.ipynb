```python
from ultralytics import YOLO
```

```python
model = YOLO('yolo11n.pt') 

results = model.train(
    data='../dataset/data.yaml',     
    epochs=50,            
    imgsz=640,            
    batch=16,            
    name='fire_smoke_v11_aug',  
    patience=10,            
    device=0,
    
    hsv_h=0.015, 
    hsv_s=0.9,    
    hsv_v=0.6,
    mosaic=1.0
)
```
